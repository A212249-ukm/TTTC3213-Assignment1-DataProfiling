## 2. Data Source and Acquisition

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)
df_original = df.copy()

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Dataset information:")
df.info()

print("\nSummary statistics:")
display(df.describe(include="all").T)

print("\nColumn data types:")
display(df.dtypes)
```

## 3. Missing Value Analysis

Missing values can affect the accuracy and completeness of subsequent analysis.

The number and percentage of missing values will be calculated for every column. Different handling methods will be applied depending on the meaning of each variable.

```python
missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

display(missing.sort_values(
    "Missing Count", ascending=False
))
```

```python
missing_pct = df.isnull().mean() * 100
missing_pct = missing_pct[missing_pct > 0]

plt.figure(figsize=(9, 5))
missing_pct.sort_values().plot(kind="barh")

plt.title("Missing Values by Column")
plt.xlabel("Missing Values (%)")
plt.ylabel("Columns")
plt.tight_layout()
plt.show()
```


**Interpretation:** The chart identifies columns with incomplete information, highlighting where missing-value treatment or explicit unknown categories are required.


### Missing Value Treatment

- **Age:** Missing ages will be replaced using the median age within each passenger class and sex group. This preserves differences between passenger groups better than using a single overall median.
- **Embarked:** Missing values will be replaced using the most frequent embarkation port, because this is a categorical variable with relatively few missing records.
- **Embark Town:** Missing values will be filled consistently using the corresponding embarkation code.
- **Deck:** The column has substantial missing information. Missing deck values will be labelled "Unknown" instead of inventing deck assignments.

The original dataset is retained separately so that the cleaning decisions can be audited.

```python

df_clean = df.copy()

# Fill age using median within passenger class and sex
df_clean["age"] = df_clean["age"].fillna(
    df_clean.groupby(["pclass", "sex"])["age"].transform("median")
)

# Fallback if any age is still missing
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())

# Fill missing embarked values with mode
embarked_mode = df_clean["embarked"].mode()[0]
df_clean["embarked"] = df_clean["embarked"].fillna(embarked_mode)

# Ensure embark_town matches embarked
port_mapping = {
    "C": "Cherbourg",
    "Q": "Queenstown",
    "S": "Southampton"
}

df_clean["embark_town"] = df_clean["embarked"].map(port_mapping)

# Preserve missing deck as Unknown
df_clean["deck"] = df_clean["deck"].cat.add_categories(
    ["Unknown"]
).fillna("Unknown") if isinstance(
    df_clean["deck"].dtype, pd.CategoricalDtype
) else df_clean["deck"].fillna("Unknown")

print("Missing values after treatment:")
display(df_clean.isnull().sum())
```


## 4. Outlier Detection and Treatment

The Interquartile Range (IQR) method is used to detect unusual fare values.

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

An unusually high fare is not automatically an error. It may represent a legitimate expensive ticket. Therefore, valid positive fare outliers will be retained and flagged for further investigation rather than deleted.


```python

Q1 = df_clean["fare"].quantile(0.25)
Q3 = df_clean["fare"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_clean["fare_outlier"] = (
    (df_clean["fare"] < lower_bound) |
    (df_clean["fare"] > upper_bound)
)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of fare outliers:",
      df_clean["fare_outlier"].sum())

display(
    df_clean.loc[df_clean["fare_outlier"],
                 ["pclass", "fare", "age"]].head(10)
)
```

```python

plt.figure(figsize=(9, 5))

sns.boxplot(data=df_clean, x="pclass", y="fare")

plt.title("Fare Distribution and Potential Outliers by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()
```


**Interpretation:** The boxplot highlights potentially extreme fare values across passenger classes. These values require investigation but are not automatically treated as errors.



### Outlier Treatment Decision

Fare outliers were detected using the IQR method and visualised using a boxplot.

These observations were not automatically removed because higher fares may be valid for premium passenger classes or group bookings.

The `fare_outlier` column records which observations require further investigation. This avoids losing potentially legitimate passenger records.

However, the IQR method may flag valid values when the fare distribution is strongly skewed. Therefore, the results should be interpreted with business context.



## 5. Other Data Quality Issues

The dataset will be checked for:

1. Duplicate records.
2. Incorrect data types.
3. Invalid age and fare values.
4. Inconsistent embarkation codes.
5. Redundant or logically inconsistent categorical columns.

Because the dataset does not contain a unique passenger identifier, exact duplicate rows cannot automatically be assumed to represent duplicate passengers.


```python

print("Exact duplicate rows:",
      df_clean.duplicated().sum())

print("Negative ages:",
      (df_clean["age"] < 0).sum())

print("Negative fares:",
      (df_clean["fare"] < 0).sum())

print("Invalid passenger classes:",
      (~df_clean["pclass"].isin([1, 2, 3])).sum())

print("Unexpected embarkation codes:",
      (~df_clean["embarked"].isin(["C", "Q", "S"])).sum())

print("\nData types:")
display(df_clean.dtypes)

print("\nMissing values:")
display(df_clean.isnull().sum())
```

```python

# Flag invalid numeric values without silently deleting records
df_clean["invalid_age"] = (
    (df_clean["age"] < 0) |
    (df_clean["age"] > 120)
)

df_clean["invalid_fare"] = df_clean["fare"] < 0

# Convert embarkation codes to consistent string type
df_clean["embarked"] = df_clean["embarked"].astype("string")

# Check consistency of survival columns
survival_mapping = {0: "no", 1: "yes"}

survival_mismatch = (
    df_clean["survived"].map(survival_mapping)
    != df_clean["alive"].astype(str)
)

print("Invalid age records:", df_clean["invalid_age"].sum())
print("Invalid fare records:", df_clean["invalid_fare"].sum())
print("Survival label mismatches:", survival_mismatch.sum())

print("Exact duplicate rows retained:",
      df_clean.duplicated().sum())
```


### Data Quality Treatment Decisions

- **Duplicates:** Exact duplicate rows are reported but retained because there is no unique passenger ID to confirm whether they are duplicate records or different passengers with identical attributes.
- **Invalid values:** Negative fares and implausible ages are flagged for investigation. Records are not removed without evidence.
- **Data types:** Embarkation codes are converted to a consistent string representation.
- **Categorical consistency:** Embarkation town names are standardised using their corresponding codes.
- **Redundant information:** Columns such as `survived` and `alive` represent the same outcome in different formats. Their consistency is checked before downstream use.

These decisions prioritise data traceability and avoid unnecessary loss of information.


```python

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df_clean,
    x="age",
    bins=25
)

plt.title("Age Distribution After Missing Value Treatment")
plt.xlabel("Age (Years)")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()
```


**Interpretation:** This chart shows the age distribution after imputation, but it does not independently establish whether the imputed values are accurate.



## 6. Privacy, PII and Bias Risks

### Privacy and PII Risks

The dataset contains demographic and travel-related attributes such as age, sex, passenger class, fare, and embarkation location.

Although the dataset does not provide passenger names or direct contact details, combinations of demographic attributes may create re-identification risks when linked with external historical records.

**Mitigation:**
- Limit access to raw passenger-level data.
- Use aggregated statistics when publishing results.
- Avoid combining records with external identifying information without a valid purpose.

### Bias Risks

The dataset represents passengers from one historical maritime disaster rather than a representative sample of the general population.

Passenger class, sex, age, and survival status reflect historical social conditions. Missing age information may also affect certain passenger groups differently.

**Mitigation:**
- Compare missing value rates across passenger classes and sex groups.
- Avoid generalising findings beyond the Titanic population.
- Document how imputation may affect group-level comparisons.
- Avoid treating historical survival differences as universal patterns.


```python

age_missing_by_group = (
    df_original.groupby(["pclass", "sex"])["age"]
    .apply(lambda x: x.isna().mean() * 100)
    .reset_index(name="Missing Age (%)")
)

display(age_missing_by_group)
```


## 7. Before and After Data Quality Comparison

This section compares the original dataset with the cleaned dataset to evaluate the impact of the cleaning process.

Missing values are assessed before and after treatment, while outliers and potentially duplicated records are reported separately to avoid deleting legitimate information.


```python

comparison = pd.DataFrame({
    "Column": df_original.columns,
    "Missing Before": df_original.isna().sum().values,
    "Missing After": df_clean[df_original.columns].isna().sum().values
})

comparison["Values Treated"] = (
    comparison["Missing Before"] -
    comparison["Missing After"]
)

display(comparison)

print("Total missing before:",
      df_original.isna().sum().sum())

print("Total missing after:",
      df_clean[df_original.columns].isna().sum().sum())
```


## 8. Duplicate Record Investigation

Exact duplicate records are investigated to determine whether they should be removed.

Because the dataset does not contain a unique passenger identifier, identical rows may represent different passengers with matching attributes. Therefore, potential duplicates are retained unless additional evidence confirms that they are erroneous duplicates.


```python

duplicate_mask = df_original.duplicated(keep=False)

print("Duplicate rows excluding first occurrence:",
      df_original.duplicated().sum())

print("Rows belonging to duplicate groups:",
      duplicate_mask.sum())

display(
    df_original.loc[duplicate_mask]
    .sort_values(["pclass", "sex", "age", "fare"])
    .head(15)
)
```


## 9. Missing Data Bias Investigation

Missing data may not be distributed equally across passenger groups.

To investigate this risk, the percentage of missing age values is compared across passenger classes and sex groups using the original dataset before imputation.

Differences between groups may indicate that the missingness mechanism is not completely random, potentially affecting the reliability of demographic comparisons.


```python

bias_table = (
    df_original.groupby(["pclass", "sex"])["age"]
    .apply(lambda x: x.isna().mean() * 100)
    .reset_index(name="Missing Age Percentage")
)

display(bias_table)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=bias_table,
    x="pclass",
    y="Missing Age Percentage",
    hue="sex"
)

plt.title("Missing Age Percentage by Passenger Class and Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Missing Age (%)")
plt.legend(title="Sex")
plt.tight_layout()
plt.show()
```


**Interpretation:** This chart compares missing age percentages between passenger groups. Unequal missingness may introduce bias into age-related analysis, so group-based imputation should be interpreted cautiously.
